```python
FILENAME = "todo_list.txt"

def load_todos():
    """Loads all To-Do entries from the file."""
    try:
        with open(FILENAME, 'r') as f:

            return [line.strip() for line in f.readlines()]
    except FileNotFoundError:

        return []

def save_todos(todos):
    """Writes the current list of To-Do entries back to the file (overwrites)."""
    with open(FILENAME, 'w') as f:

        for todo in todos:
            f.write(todo + '\n')

def add_todo():
    """Adds a new pending To-Do item."""
    description = input("Enter new To-Do description: ").strip()
    if not description:
        print("Description cannot be empty.")
        return

    todos = load_todos()

    new_todo = f"[ ] {description}"
    todos.append(new_todo)
    save_todos(todos)
    print("Task added successfully.")

def list_todos():
    """Displays all To-Do items with their index."""
    todos = load_todos()
    if not todos:
        print("\n*** The To-Do list is currently empty. ***")
        return

    print("\n--- Your To-Do List ---")
    for i, todo in enumerate(todos):

        print(f"{i+1}. {todo}")
    print("-------------------------")

def mark_complete():
    """Marks a To-Do item as complete."""
    list_todos()
    if not load_todos():
        return

    try:
        task_num = int(input("Enter the number of the task to mark as COMPLETE: "))
        todos = load_todos()


        if 1 <= task_num <= len(todos):
            index = task_num - 1
            todo_item = todos[index]


            if todo_item.startswith('[x]'):
                print("Task is already complete.")
                return


            todos[index] = todo_item.replace('[ ]', '[x]', 1)
            save_todos(todos)
            print(f"Task {task_num} marked as complete: {todos[index]}")
        else:
            print("Invalid task number.")

    except ValueError:
        print("Invalid input. Please enter a number.")

def delete_todo():
    """Deletes a To-Do item."""
    list_todos()
    if not load_todos():
        return

    try:
        task_num = int(input("Enter the number of the task to DELETE: "))
        todos = load_todos()


        if 1 <= task_num <= len(todos):
            index = task_num - 1
            deleted_todo = todos.pop(index)
            save_todos(todos)
            print(f"Task deleted: {deleted_todo}")
        else:
            print("Invalid task number.")

    except ValueError:
        print("Invalid input. Please enter a number.")


def main_menu():
    """Displays the main menu and handles user interaction."""
    while True:
        print("\n===== To-Do Logger =====")
        print("1. Add Task")
        print("2. List Tasks")
        print("3. Mark Task as Complete")
        print("4. Delete Task")
        print("5. Exit")
        print("------------------------")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == '1':
            add_todo()
        elif choice == '2':
            list_todos()
        elif choice == '3':
            mark_complete()
        elif choice == '4':
            delete_todo()
        elif choice == '5':
            print("Exiting To-Do Logger. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 5.")

```
